```python

from google.colab import files
import zipfile
import os

uploaded = files.upload()

zip_path = next(
    name for name in uploaded
    if name.lower().endswith(".zip")
)

extract_dir = "/content/driver_dataset"
os.makedirs(extract_dir, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_dir)

print("Dataset extracted successfully!")
print("Location:", extract_dir)

```

```python
import os

def find_folder(parent, folder_name):
    matches = []

    for root, dirs, _ in os.walk(parent):
        dirs[:] = [
            d for d in dirs
            if not d.startswith(".")
        ]

        for directory in dirs:
            if directory.upper() == folder_name:
                matches.append(os.path.join(root, directory))

    return matches

# Using extract_dir which was defined in the previous cell
train_matches = find_folder(extract_dir, "TRAIN")
val_matches = find_folder(extract_dir, "VALIDATION")
test_matches = find_folder(extract_dir, "TEST")

if not train_matches or not val_matches or not test_matches:
    raise FileNotFoundError(
        "Could not find all three folders. "
        "Check the extracted dataset structure."
    )

# Find the TRAIN folder whose parent also contains
# VALIDATION and TEST.
DATA_ROOT = None

for candidate in train_matches:
    parent = os.path.dirname(candidate)

    if (
        os.path.isdir(os.path.join(parent, "VALIDATION"))
        and os.path.isdir(os.path.join(parent, "TEST"))
    ):
        DATA_ROOT = parent
        break

if DATA_ROOT is None:
    raise ValueError(
        "TRAIN, VALIDATION and TEST must be sibling folders."
    )

TRAIN_DIR = os.path.join(DATA_ROOT, "TRAIN")
VAL_DIR = os.path.join(DATA_ROOT, "VALIDATION")
TEST_DIR = os.path.join(DATA_ROOT, "TEST")

print("TRAIN:", TRAIN_DIR)
print("VALIDATION:", VAL_DIR)
print("TEST:", TEST_DIR)
```

```python
import os

IMAGE_EXTENSIONS = (
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
)

def get_classes(folder):
    return sorted([
        name for name in os.listdir(folder)
        if os.path.isdir(os.path.join(folder, name))
        and not name.startswith(".")
    ])

train_classes_raw = get_classes(TRAIN_DIR)
val_classes_raw = get_classes(VAL_DIR)
test_classes_raw = get_classes(TEST_DIR)

# Map split-specific names to a unified, clean class name (e.g., 'EYE_TRAIN' -> 'EYE')
def clean_class_name(name):
    for suffix in ["_TRAIN", "_VAL", "_TEST"]:
        if name.endswith(suffix):
            return name[:-len(suffix)]
    return name

train_classes = sorted(list(set(clean_class_name(c) for c in train_classes_raw)))
val_classes = sorted(list(set(clean_class_name(c) for c in val_classes_raw)))
test_classes = sorted(list(set(clean_class_name(c) for c in test_classes_raw)))

print("Normalized Training classes:", train_classes)
print("Normalized Validation classes:", val_classes)
print("Normalized Test classes:", test_classes)

assert len(train_classes) >= 2, "Need at least two classes."
assert train_classes == val_classes == test_classes, (
    "Normalized class names must match in all three splits."
)

def count_images(folder, raw_classes):
    counts = {}
    for raw_name in raw_classes:
        clean_name = clean_class_name(raw_name)
        class_dir = os.path.join(folder, raw_name)
        counts[clean_name] = sum(
            1 for f in os.listdir(class_dir)
            if f.lower().endswith(IMAGE_EXTENSIONS)
        )
    return counts

print("\nTRAIN:", count_images(TRAIN_DIR, train_classes_raw))
print("VALIDATION:", count_images(VAL_DIR, val_classes_raw))
print("TEST:", count_images(TEST_DIR, test_classes_raw))
```

```python

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import time

IMG_SIZE = (96, 96)
BATCH_SIZE = 32
SEED = 42

NUM_CLASSES = len(train_classes)

# We pass the raw folder names that match the subdirectories
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    class_names=train_classes_raw,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    class_names=val_classes_raw,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    class_names=test_classes_raw,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=False
)

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

print("Class mapping:", dict(enumerate(train_classes)))
print("Number of classes:", NUM_CLASSES)
```

```python

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import time

IMG_SIZE = (96, 96)
BATCH_SIZE = 32
SEED = 42

NUM_CLASSES = len(train_classes)

# We must use the split-specific raw folder names for disk directories
train_ds = tf.keras.utils.image_dataset_from_directory(
    TRAIN_DIR,
    class_names=train_classes_raw,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=True,
    seed=SEED
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    VAL_DIR,
    class_names=val_classes_raw,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    TEST_DIR,
    class_names=test_classes_raw,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int",
    shuffle=False
)

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
val_ds = val_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

print("Class mapping:", dict(enumerate(train_classes)))
print("Number of classes:", NUM_CLASSES)
```

```python

from tensorflow.keras import layers, models
import tensorflow as tf
import time
import os

cnn = models.Sequential([
    layers.Input(shape=(96, 96, 3)),
    layers.Rescaling(1.0 / 255),

    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),

    layers.Conv2D(32, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),

    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.4),

    layers.Dense(NUM_CLASSES, activation="softmax")
])

cnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn.summary()

start = time.time()

cnn_history = cnn.fit(
    train_ds,
    validation_data=val_ds
)

cnn_time = time.time() - start

os.makedirs("/content/models", exist_ok=True)
cnn.save("/content/models/custom_cnn.keras")

print("CNN training time:", round(cnn_time, 2), "seconds")
print("CNN model saved!")
```

```python

from tensorflow.keras import layers, models
import tensorflow as tf
import time
import os

cnn = models.Sequential([
    layers.Input(shape=(96, 96, 3)),
    layers.Rescaling(1.0 / 255),

    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.05),

    layers.Conv2D(32, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, padding="same", activation="relu"),
    layers.MaxPooling2D(),

    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation="relu"),
    layers.Dropout(0.4),

    layers.Dense(NUM_CLASSES, activation="softmax")
])

cnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn.summary()

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    )
]

start = time.time()

cnn_history = cnn.fit(
    train_ds,
    validation_data=val_ds,
    epochs=3,
    callbacks=callbacks
)

cnn_time = time.time() - start

os.makedirs("/content/models", exist_ok=True)
cnn.save("/content/models/custom_cnn.keras")

print("CNN training time:", round(cnn_time, 2), "seconds")
print("CNN model saved successfully!")
```

```python

from tensorflow.keras.applications import MobileNetV2
import os
import time
from tensorflow.keras import layers, models

os.makedirs("/content/models", exist_ok=True)

base_model = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(96, 96, 3)
)

# Freeze pretrained layers
base_model.trainable = False

inputs = layers.Input(shape=(96, 96, 3))

# Preprocess pixels from 0–255 to -1–1
x = layers.Rescaling(1.0 / 127.5, offset=-1)(inputs)

x = base_model(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

outputs = layers.Dense(
    NUM_CLASSES,
    activation="softmax"
)(x)

mobilenet = models.Model(inputs, outputs)

mobilenet.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

mobilenet.summary()

start = time.time()

mobilenet_history = mobilenet.fit(
    train_ds,
    validation_data=val_ds,
    epochs=3,  # Reduced from 10 to 3
    callbacks=callbacks
)

mobilenet_time = time.time() - start

mobilenet.save("/content/models/mobilenetv2.keras")

print("MobileNetV2 training time:", round(mobilenet_time, 2), "seconds")
print("MobileNetV2 model saved!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    precision_recall_fscore_support
)

def evaluate_model(model, dataset, model_name):
    y_true = np.concatenate([
        labels.numpy()
        for _, labels in dataset
    ])

    probabilities = model.predict(dataset, verbose=0)
    y_pred = np.argmax(probabilities, axis=1)

    accuracy = accuracy_score(y_true, y_pred)

    precision, recall, f1, _ = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )
    )

    print("\n", "=" * 50)
    print(model_name)
    print("=" * 50)
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-score:  {f1:.4f}")

    print("\nClassification Report:")
    print(classification_report(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES)),
        target_names=train_classes,
        zero_division=0
    ))

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=list(range(NUM_CLASSES))
    )

    display = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=train_classes
    )

    display.plot(values_format="d", xticks_rotation=45)
    plt.title(model_name + " Test Confusion Matrix")
    plt.show()

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1
    }

cnn_results = evaluate_model(cnn, test_ds, "Custom CNN")
mobilenet_results = evaluate_model(
    mobilenet, test_ds, "MobileNetV2"
)
```
